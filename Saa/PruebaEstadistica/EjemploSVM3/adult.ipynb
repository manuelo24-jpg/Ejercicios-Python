```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
df_adults = pd.read_csv('adult.csv')
df_adults = pd.get_dummies(df_adults, drop_first=True, columns=['income'], dtype='int')
categoric_columns = df_adults.select_dtypes(include=['object']).columns

print(df_adults.info())
print(df_adults.shape)
df_adults.replace('?', np.nan, inplace=True)
df_adults.dropna(inplace=True)
print(df_adults.shape)
df_adults.drop(columns=categoric_columns, axis=1,inplace=True)
```

```python
X = df_adults.drop(['income_>50K'], axis=1)
y = df_adults['income_>50K'].to_frame()

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
svm_lineal = SVC(kernel="linear", C=1.0, random_state=42)
svm_lineal.fit(X_train_scaled, y_train)

pred_lineal = svm_lineal.predict(X_test_scaled)

print("=== SVM LINEAL ===")
print("Accuracy:", accuracy_score(y_test, pred_lineal)) #exactitud
print(confusion_matrix(y_test, pred_lineal))
print(classification_report(y_test, pred_lineal))
```

```python
"""svm_rbf = SVC(kernel="rbf", C=2.0, gamma=1.0, random_state=42)
svm_rbf.fit(X_train_scaled, y_train)"""
valores_C = [0.1, 1, 4, 7, 10]
valores_gamma = [0.001, 0.01, 0.1, 1, 3, 5]

mejor_acc = 0
mejor_config = None

for C in valores_C:
    for gamma in valores_gamma:
        print(f"\nProbando C={C} | gamma={gamma}")
        
        svm_rbf = SVC(kernel="rbf", C=C, gamma=gamma, random_state=42)
        svm_rbf.fit(X_train_scaled, y_train)

        pred = svm_rbf.predict(X_test_scaled)
        acc = accuracy_score(y_test, pred)

        print(f"Accuracy: {acc:.4f}")

        if acc > mejor_acc:
            mejor_acc = acc
            mejor_config = (C, gamma)

pred_rbf = svm_rbf.predict(X_test_scaled)

print(f"Mejor Accuracy: {mejor_acc:.4f}")
print(f"Mejor C y gamma: C={mejor_config[0]}, gamma={mejor_config[1]}")
print(confusion_matrix(y_test, pred_rbf))
print(classification_report(y_test, pred_rbf))
```

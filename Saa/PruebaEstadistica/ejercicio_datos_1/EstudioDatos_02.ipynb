# Análisis de Datos con el Dataset "Adult Income"

El Dataset "Adult Income" es una colección de datos demográficos y laborales extraídos del censo estadounidense de 1994, recopilado por la Oficina del Censo de los Estados Unidos. Este conjunto de datos es ampliamente utilizado en tareas de aprendizaje automático, estadística y análisis de datos para estudios relacionados con clasificación, regresión y pruebas de hipótesis estadísticas.

El objetivo principal asociado con este dataset es predecir si un individuo gana más de $50,000 al año en función de sus atributos personales y laborales. Este análisis permite explorar relaciones entre variables y realizar interpretaciones significativas a nivel social y económico.

## Descripción de las Variables

El dataset incluye las siguientes variables:

- **age**: Edad del individuo (en años).
- **workclass**: Tipo de empleo (p. ej., privado, gubernamental, autónomo, etc.).
- **fnlwgt**: Peso final asignado a la observación, utilizado para expandir la muestra del censo.
- **education**: Nivel educativo alcanzado (p. ej., secundaria, licenciatura, maestría).
- **education.num**: Nivel educativo en formato numérico (años de educación).
- **marital.status**: Estado civil (p. ej., casado, soltero, divorciado).
- **occupation**: Ocupación laboral (p. ej., administrativo, técnico, etc.).
- **relationship**: Relación familiar con el cabeza de familia (p. ej., esposo/a, hijo/a, etc.).
- **race**: Raza del individuo (p. ej., blanca, negra, asiática).
- **sex**: Género del individuo (masculino o femenino).
- **capital.gain**: Ganancia de capital obtenida.
- **capital.loss**: Pérdida de capital reportada.
- **native.country**: País de origen.
- **hours.per.week**: Número de horas trabajadas por semana.
- **income**: Variable objetivo, indica si el ingreso anual es '>50K' o '<=50K'.

```python
import pandas as pd
import numpy as np
import statistics
import matplotlib.pyplot as plt
from pathlib import Path

outliersList = []

# Intentar leer el fichero desde varias rutas comunes y dar un mensaje claro si no se encuentra
csv_candidates = [
    "adult.csv",
    "adult.data",
    "./adult.csv",
    "../adult.csv"
]

df_adults = None
for p in csv_candidates:
    if Path(p).exists():
        try:
            df_adults = pd.read_csv(p, skipinitialspace=True)
            break
        except Exception as e:
            print(f"No se pudo leer {p}: {e}")

if df_adults is None:
    raise FileNotFoundError(
        "No se encontró el archivo 'adult.csv' (ni alternativas). "
        "Coloque el archivo en el directorio de trabajo o actualice la ruta."
    )

# Información general del dataset
df_adults.info()
print("El dataset tiene {} filas y {} columnas".format(df_adults.shape[0], df_adults.shape[1]))
print("En vista de que este dataset tiene el objetivo de predecir si una persona " \
"tiene un salario superior a 50K, creo que tiene datos bastante interesantes y completos")

# Cálculo de estadísticas descriptivas
print("La media de edad de los adultos es {}".format(np.mean(df_adults['age'])))
print("La mediana de edad de los adultos es {}".format(np.median(df_adults['age'])))

# Usar pandas.Series.mode() para evitar error si hay múltiples modas
mode_vals = df_adults['capital.gain'].mode()
mode_capital_gain = mode_vals.iloc[0] if not mode_vals.empty else np.nan
print("La moda de capital ganado de los adultos es {}".format(mode_capital_gain))

# Varianza y desviación estándar (utilizando numpy para robustez)
print("La varianza de capital perdido de los adultos es {}".format(np.var(df_adults['capital.loss'], ddof=1)))
print("La desviación estándar de capital perdido de los adultos es {}".format(np.std(df_adults['capital.loss'], ddof=1)))

# Visualización de datos
df_adults.sort_values(by='age', inplace=True)
plt.figure(figsize=(8,4))
plt.plot(df_adults['age'], df_adults['capital.gain'], marker='.', linestyle='none', alpha=0.6)
plt.title("Grafico de capital ganado por edad")
plt.xlabel("Edad")
plt.ylabel("Capital Ganado")
plt.tight_layout()
plt.show()

plt.figure(figsize=(6,4))
plt.hist(df_adults['hours.per.week'], bins=20)
plt.title("Histograma de horas trabajadas por semana")
plt.xlabel("Horas por semana")
plt.ylabel("Frecuencia")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8,6))
edu_counts = df_adults['education'].value_counts()
plt.barh(edu_counts.index, edu_counts.values)
plt.title("Gráfico de barras horizontales de nivel educativo")
plt.xlabel("Número de personas")
plt.ylabel("Nivel educativo")
plt.tight_layout()
plt.show()

plt.figure(figsize=(6,4))
plt.scatter(df_adults['age'], df_adults['capital.loss'], alpha=0.5)
plt.title("Grafico de capital perdido por edad")
plt.xlabel("Edad")
plt.ylabel("Capital perdido")
plt.tight_layout()
plt.show()

ct = pd.crosstab(df_adults['sex'], df_adults['income'])
ct.plot(kind='bar')
plt.title("Distribución de ingresos por sexo")
plt.xlabel("Sexo")
plt.ylabel("Número de personas")
plt.xticks(rotation=0)
plt.legend(title='Income')
plt.tight_layout()
plt.show()

plt.figure(figsize=(6,6))
df_adults["education"].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Distribución de nivel educativo")
plt.ylabel("")
plt.tight_layout()
plt.show()

plt.figure(figsize=(6,6))
df_adults["income"].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Distribución de ingresos")
plt.ylabel("")
plt.tight_layout()
plt.show()


#Detectar valores atípicos
# Usar selección vectorizada en lugar de iterrows para obtener ganancias > 0
adults_with_gains = df_adults.loc[df_adults['capital.gain'] > 0, 'capital.gain'].tolist()

if len(adults_with_gains) == 0:
    print("No hay registros con capital.gain > 0")
else:
    percentiles = np.percentile(adults_with_gains, [25, 50, 75])
    print("Cuartiles: {}".format(percentiles))
    q1 = percentiles[0]
    q3 = percentiles[2]
    iqr = q3 - q1

    print("El rango intercuartilico es {}".format(iqr))

    outliersList = [v for v in adults_with_gains if (v < (q1 - 1.5 * iqr) or v > (q3 + 1.5 * iqr))]
    print("Los valores atipicos son {} ".format(outliersList))

# Relaciones entre variables
print(df_adults.corr(numeric_only=True)['age'].abs().sort_values())
print(df_adults.corr(numeric_only=True)['capital.gain'].abs().sort_values())
print(df_adults.corr(numeric_only=True)['hours.per.week'].abs().sort_values())

```

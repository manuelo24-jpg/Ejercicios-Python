# Métodos para el cálculo de distancias

Todos los métodos de clustering necesitan definir y cuantificar la similitud entre las observaciones, para ello se utiliza el concepto de **distancia**.

En este notebook vamos a ver diferentes métodos para el cálculo de distancias.

## Distancia euclídea

La distancia euclídea entre dos puntos se define como la longitud del segmento que los une.
<img src="https://upload.wikimedia.org/wikipedia/commons/thumb/6/67/Distance_Formula.svg/1200px-Distance_Formula.svg.png" width="700">

Este cálculo se puede ampliar a n dimensiones.

En clustering es común utilizar la distancia eucídea al cuadrado para asignar mayor peso a los datos alejados.

## Distancia de Manhattan

Se define la distancia entre dos puntos  p  y  q  como el sumatorio de las diferencias absolutas entre cada dimensión. Esta medida se ve menos afectada por outliers (es más robusta) que la distancia euclídea debido a que no eleva al cuadrado las diferencias.

Comparación entre la distancia euclídea (verde) y la distancia de manhattan (segmento rojo, amarillo y azul) en un espacio bidimensional. Existen múltiples caminos para unir dos puntos con una misma distancia de manhattan, pero solo uno con la distancia euclídea.

<img src="https://upload.wikimedia.org/wikipedia/commons/thumb/0/08/Manhattan_distance.svg/1200px-Manhattan_distance.svg.png" width="500">

## Correlación

La correlación determina la distancia entre dos muestras teniendo en cuenta patrones en lugar del desplazamiento de una magnitud.

Ejemplo: obs_1 y obs_2 tienen el mismo patrón, pero obs_2 está desplazada por debajo de obs_1. obs_3 tiene un patrón distinto a las demás, pero su rango de valores es similar al de obs_2. Acorde a la distancia euclídea, las observaciones obs_2 y obs_3 son las más similares ya que la suma de la distancia al cuadrado entre sus valores es menor. Sin embargo, **acorde a la correlación de Pearson, las observaciones más similares son obs_1 y obs_2 ya que siguen el mismo patrón**.

# Cálculo de distancias en Aprendizaje No Supervisado

En aprendizaje no supervisado, los algoritmos trabajan sin etiquetas.  
Por eso, medir la **similitud o diferencia entre datos** es fundamental.

Las métricas más utilizadas son:

- **Distancia euclídea**: distancia "en línea recta"
- **Distancia Manhattan**: suma de diferencias absolutas
- **Distancia Minkowski**: generalización que incluye las dos anteriores
Estas distancias se utilizan en algoritmos como **K-Means** o **clustering jerárquico**.

```python
# IMportación de Librerías
import pandas as pd
from sklearn.metrics import pairwise_distances
from sklearn.preprocessing import scale
```

```python
# Carga de Datos
# Dataset USArrests
url = "https://raw.githubusercontent.com/vincentarelbundock/Rdatasets/master/csv/datasets/USArrests.csv"
datos = pd.read_csv(url, index_col=0)

datos.head()
```

## Escalado de Datos

```python
# Escalado (muy importante en clustering)
datos_scaled = scale(datos)

# Convertimos a DataFrame para mantener nombres
datos_scaled = pd.DataFrame(datos_scaled, index=datos.index, columns=datos.columns)

datos_scaled.head()
```

## Distancia Euclídea

```python
dist_euclidea = pairwise_distances(datos_scaled, metric='euclidean')

dist_euclidea = pd.DataFrame(dist_euclidea, index=datos.index, columns=datos.index)

dist_euclidea.head()
```

## Distancia Manhattan

```python
dist_manhattan = pairwise_distances(datos_scaled, metric='manhattan')

dist_manhattan = pd.DataFrame(dist_manhattan, index=datos.index, columns=datos.index)

dist_manhattan.head()
```

## Búsqueda de estados similares

```python
# Nos quedamos con la mitad de la matriz (sin duplicados)
import numpy as np

indices = np.triu_indices(dist_euclidea.shape[0], k=1)


distancias = dist_euclidea.values[indices]

pares = list(zip(indices[0], indices[1], distancias))

pares_ordenados = sorted(pares, key=lambda x: x[2])

for i in range(5):
    a = datos.index[pares_ordenados[i][0]]
    b = datos.index[pares_ordenados[i][1]]
    d = pares_ordenados[i][2]
    print(f"{a} - {b}: {d:.2f}")
```

## Búsqueda de estados menos similares

```python
# Los más lejanos
for i in range(1,6):
    a = datos.index[pares_ordenados[-i][0]]
    b = datos.index[pares_ordenados[-i][1]]
    d = pares_ordenados[-i][2]
    print(f"{a} - {b}: {d:.2f}")
```

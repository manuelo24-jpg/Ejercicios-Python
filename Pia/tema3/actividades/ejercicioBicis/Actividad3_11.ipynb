```python
from sklearn.datasets import make_moons
from sklearn.tree import DecisionTreeRegressor
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

from sklearn.tree import export_graphviz
from sklearn.datasets import make_circles
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, mean_squared_error, mean_absolute_error, r2_score
```

```python
import pandas as pd
df_bikes = pd.read_csv("SeoulBikeData.csv", sep=',',encoding='latin-1')
df_bikes.head()
```

```python
df_bikes.drop(columns=["Date"], axis=1, inplace=True)
df_bikes = pd.get_dummies(df_bikes, columns=["Seasons", "Holiday", "Functioning Day"], dtype=int)
```

```python
df_bikes.head()
```

```python
df_bikes.corr(numeric_only=True)['Rented Bike Count'].abs().sort_values(ascending=False)[1:]
```

```python
#Generamos los conjuntos X e y.
X = df_bikes.drop(['Rented Bike Count'],axis=1)
y = df_bikes['Rented Bike Count'].to_frame()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

```python
tree_clf = DecisionTreeRegressor(max_depth=9, min_samples_leaf=2, min_samples_split=20)
tree_clf.fit(X_train, y_train)

# For a regressor there is no `classes_` attribute, so omit `class_names`.
export_graphviz(
    tree_clf,
    out_file="./seul_bike.dot",
    feature_names=list(df_bikes.drop(['Rented Bike Count'], axis=1).columns),
    rounded=True,
    filled=True
)

# If "dot: command not found" occurs, install graphviz in the system:
# sudo apt install graphviz
!dot -Tpng seul_bike.dot -o seul_bike.png
```

```python
y_pred = tree_clf.predict(X_test)
print(y_pred)
print("Mean Absolute Error:", mean_absolute_error(y_test, y_pred))
print("R^2 Score:", tree_clf.score(X_test, y_test))
print("Mean Squared Error:", mean_squared_error(y_test,y_pred))
```

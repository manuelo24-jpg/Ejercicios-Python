```python
# Instala la librería necesaria si no está disponible
%pip install -q transformers

from transformers import AutoTokenizer

# 1. Definición del texto de entrada
prompt = """Yeah Perdonen, Kamehameha
Después del tema del Tetris viene el Dragon Ball Rap
Quien no haya seguido esta serie es porque no tiene infancia
Big Bang Attack ataca desde el Planeta Namek"""  

# 2. Carga del tokenizador preentrenado (Qwen 2 con 500M de parámetros)
tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen2-0.5B")

# 3. Conversión del texto a identificadores numéricos (input_ids)
input_ids = tokenizer(prompt).input_ids

# 4. Iteración y decodificación para visualizar la tokenización
print(f"{'ID':<10} | {'Token Decodificado'}")
print("-" * 30)

for t in input_ids:
    token_texto = tokenizer.decode(t)
    print(f"{t:<10} | {token_texto}")
```

```python
from os import listdir
from numpy import asarray
from numpy import save
import tensorflow as tf
#Deshabilitar la GPU:
#tf.config.set_visible_devices([], 'GPU')
from tensorflow import keras
from tensorflow.keras.utils import load_img
from tensorflow.keras.utils import img_to_array

photos =  []
image = []
```

```python
folders = listdir('img_align_celeba/img_align_celeba')

len(folders)

for idx, folder in enumerate(folders):

    for file in listdir('img_align_celeba/img_align_celeba' )[:1000]:
        photo = load_img('img_align_celeba/img_align_celeba/'+folder, target_size=(50,50))
        photo = img_to_array(photo)
        image.append(folder)
        photos.append(photo)
```

```python
import pandas as pd


df_famosos = pd.read_csv('list_attr_celeba.csv')
df_famosos.head()
```

```python
lista_atractivos = []
lista_calvos = []
lista_bolsas = []
```

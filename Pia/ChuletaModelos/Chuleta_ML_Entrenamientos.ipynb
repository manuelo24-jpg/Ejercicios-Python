## 📦 IMPORTS COMPLETOS

```python
# Datos y preprocesamiento
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV, cross_val_score
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# Modelos de Clasificación
from sklearn.svm import SVC, LinearSVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# Modelos de Regresión
from sklearn.svm import SVR, LinearSVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

# Métricas de Clasificación
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Métricas de Regresión
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

# Visualización
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración
import warnings
warnings.filterwarnings('ignore')
plt.style.use('seaborn-v0_8-darkgrid')

print("✅ Todas las librerías importadas correctamente")
```

## 📊 GENERAR DATOS DE EJEMPLO

```python
from sklearn.datasets import make_classification, make_regression

# Datos de CLASIFICACIÓN
X_clf, y_clf = make_classification(
    n_samples=1000,
    n_features=20,
    n_informative=15,
    n_redundant=5,
    n_classes=2,
    random_state=42
)

# Datos de REGRESIÓN
X_reg, y_reg = make_regression(
    n_samples=1000,
    n_features=20,
    n_informative=15,
    noise=10,
    random_state=42
)

print(f"📊 Datos de Clasificación: {X_clf.shape}")
print(f"📈 Datos de Regresión: {X_reg.shape}")
```

---
# 🔵 SVM - SUPPORT VECTOR MACHINE
---

## 📊 SVM CLASIFICACIÓN - Básico

```python
# --- PREPARAR DATOS ---
X_train, X_test, y_train, y_test = train_test_split(
    X_clf, y_clf, test_size=0.2, random_state=42
)

# --- NORMALIZAR (OBLIGATORIO PARA SVM) ---
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# --- ENTRENAR SVM ---
svm_clf = SVC(
    kernel='rbf',         # Kernel gaussiano
    C=1.0,                # Regularización
    gamma='scale',        # Coeficiente kernel
    random_state=42
)

svm_clf.fit(X_train_scaled, y_train)

# --- PREDECIR ---
y_pred = svm_clf.predict(X_test_scaled)

# --- EVALUAR ---
print("🔵 SVM CLASIFICACIÓN - RESULTADOS:")
print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

## 📊 SVM CLASIFICACIÓN - Lineal (más rápido)

```python
# --- ENTRENAR SVM LINEAL ---
svm_linear = LinearSVC(
    C=1.0,
    max_iter=1000,
    random_state=42
)

svm_linear.fit(X_train_scaled, y_train)
y_pred_linear = svm_linear.predict(X_test_scaled)

print("🔵 SVM LINEAL - RESULTADOS:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_linear):.4f}")
```

## 📊 SVM CLASIFICACIÓN - Grid Search

```python
# --- GRID SEARCH SVM ---
param_grid = {
    'C': [0.1, 1, 10, 100],
    'gamma': ['scale', 'auto', 0.001, 0.01, 0.1],
    'kernel': ['rbf', 'linear']
}

grid_svm = GridSearchCV(
    SVC(random_state=42),
    param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    verbose=1
)

print("🔍 Buscando mejores parámetros...")
grid_svm.fit(X_train_scaled, y_train)

print(f"\n✅ Mejores parámetros: {grid_svm.best_params_}")
print(f"✅ Mejor score CV: {grid_svm.best_score_:.4f}")

# --- PREDECIR CON MEJOR MODELO ---
y_pred_best = grid_svm.predict(X_test_scaled)
print(f"\n🎯 Accuracy Test: {accuracy_score(y_test, y_pred_best):.4f}")
```

## 📈 SVM REGRESIÓN - Básico

```python
# --- PREPARAR DATOS ---
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)

# --- NORMALIZAR ---
scaler_r = StandardScaler()
X_train_r_scaled = scaler_r.fit_transform(X_train_r)
X_test_r_scaled = scaler_r.transform(X_test_r)

# --- ENTRENAR SVR ---
svr = SVR(
    kernel='rbf',
    C=1.0,
    gamma='scale',
    epsilon=0.1
)

svr.fit(X_train_r_scaled, y_train_r)

# --- PREDECIR ---
y_pred_r = svr.predict(X_test_r_scaled)

# --- EVALUAR ---
print("🔵 SVR REGRESIÓN - RESULTADOS:")
print(f"MSE:  {mean_squared_error(y_test_r, y_pred_r):.4f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test_r, y_pred_r)):.4f}")
print(f"MAE:  {mean_absolute_error(y_test_r, y_pred_r):.4f}")
print(f"R²:   {r2_score(y_test_r, y_pred_r):.4f}")
```

## 📊 DECISION TREE CLASIFICACIÓN - Básico

```python
# --- ENTRENAR DECISION TREE ---
# NO requiere normalización
dt_clf = DecisionTreeClassifier(
    max_depth=10,
    min_samples_split=10,
    min_samples_leaf=5,
    criterion='gini',
    random_state=42
)

dt_clf.fit(X_train, y_train)

# --- PREDECIR ---
y_pred_dt = dt_clf.predict(X_test)

# --- EVALUAR ---
print("🌳 DECISION TREE CLASIFICACIÓN - RESULTADOS:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_dt):.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_dt))
```

## 📊 DECISION TREE - Grid Search

```python
# --- GRID SEARCH DECISION TREE ---
param_grid_dt = {
    'max_depth': [5, 10, 15, 20, None],
    'min_samples_split': [2, 10, 20],
    'min_samples_leaf': [1, 5, 10],
    'criterion': ['gini', 'entropy']
}

grid_dt = GridSearchCV(
    DecisionTreeClassifier(random_state=42),
    param_grid_dt,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    verbose=1
)

print("🔍 Buscando mejores parámetros...")
grid_dt.fit(X_train, y_train)

print(f"\n✅ Mejores parámetros: {grid_dt.best_params_}")
print(f"✅ Mejor score CV: {grid_dt.best_score_:.4f}")

y_pred_dt_best = grid_dt.predict(X_test)
print(f"\n🎯 Accuracy Test: {accuracy_score(y_test, y_pred_dt_best):.4f}")
```

## 📈 DECISION TREE REGRESIÓN

```python
# --- ENTRENAR DECISION TREE REGRESSOR ---
dt_reg = DecisionTreeRegressor(
    max_depth=10,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=42
)

dt_reg.fit(X_train_r, y_train_r)

# --- PREDECIR ---
y_pred_dt_reg = dt_reg.predict(X_test_r)

# --- EVALUAR ---
print("🌳 DECISION TREE REGRESIÓN - RESULTADOS:")
print(f"MSE:  {mean_squared_error(y_test_r, y_pred_dt_reg):.4f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test_r, y_pred_dt_reg)):.4f}")
print(f"MAE:  {mean_absolute_error(y_test_r, y_pred_dt_reg):.4f}")
print(f"R²:   {r2_score(y_test_r, y_pred_dt_reg):.4f}")
```

# 📊 COMPARACIÓN COMPLETA DE MODELOS
---

```python
# --- MODELOS ---
models = {
    'SVM (RBF)': SVC(kernel='rbf', C=1.0, random_state=42),
    'SVM (Linear)': LinearSVC(C=1.0, max_iter=1000, random_state=42),
    'Decision Tree': DecisionTreeClassifier(max_depth=10, random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
}

# --- ENTRENAR Y EVALUAR ---
results = {}

print("🔄 ENTRENANDO Y COMPARANDO MODELOS...\n")

for name, model in models.items():
    # Usar datos escalados solo para SVM
    if 'SVM' in name:
        X_tr, X_te = X_train_scaled, X_test_scaled
    else:
        X_tr, X_te = X_train, X_test
    
    # Cross-validation
    cv_scores = cross_val_score(model, X_tr, y_train, cv=5, scoring='accuracy')
    
    # Entrenar y test
    model.fit(X_tr, y_train)
    y_pred = model.predict(X_te)
    test_acc = accuracy_score(y_test, y_pred)
    
    results[name] = {
        'CV Mean': cv_scores.mean(),
        'CV Std': cv_scores.std(),
        'Test Acc': test_acc
    }
    
    print(f"{name}:")
    print(f"  CV: {cv_scores.mean():.4f} ± {cv_scores.std():.4f}")
    print(f"  Test: {test_acc:.4f}\n")

# --- CREAR DATAFRAME ---
df_results = pd.DataFrame(results).T
print("\n" + "="*60)
print("RESUMEN DE RESULTADOS:")
print("="*60)
print(df_results)
print("="*60)
```

## 📊 VISUALIZACIÓN DE COMPARACIÓN

```python
# --- GRÁFICO DE COMPARACIÓN ---
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Cross-Validation Scores
df_results['CV Mean'].plot(
    kind='bar',
    yerr=df_results['CV Std'],
    ax=axes[0],
    color=['#3498db', '#9b59b6', '#2ecc71', '#e74c3c']
)
axes[0].set_title('Cross-Validation Scores', fontsize=14, fontweight='bold')
axes[0].set_ylabel('Accuracy')
axes[0].set_xlabel('Modelo')
axes[0].set_xticklabels(df_results.index, rotation=45, ha='right')
axes[0].grid(axis='y', alpha=0.3)

# Test Accuracy
df_results['Test Acc'].plot(
    kind='bar',
    ax=axes[1],
    color=['#3498db', '#9b59b6', '#2ecc71', '#e74c3c']
)
axes[1].set_title('Test Accuracy', fontsize=14, fontweight='bold')
axes[1].set_ylabel('Accuracy')
axes[1].set_xlabel('Modelo')
axes[1].set_xticklabels(df_results.index, rotation=45, ha='right')
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

# --- MEJOR MODELO ---
best_model = df_results['Test Acc'].idxmax()
best_score = df_results['Test Acc'].max()

print(f"\n🏆 MEJOR MODELO: {best_model}")
print(f"🎯 Accuracy: {best_score:.4f}")
```

## ¿Qué modelo usar?

| Modelo | Normalizar | Velocidad | Mejor Para |
|--------|------------|-----------|------------|
| **SVM** | ✅ **SÍ** | Lento | Datasets pequeños, alta dimensión |
| **Decision Tree** | ❌ NO | Rápido | Interpretabilidad, datos categóricos |
| **Random Forest** | ❌ NO | Medio | Robustez, accuracy general |

## Parámetros Clave

### SVM:
- `kernel='rbf'` - Kernel gaussiano (default, recomendado)
- `C=1.0` - Regularización (0.1 = suave, 100 = estricto)
- `gamma='scale'` - Coeficiente kernel (auto)

### Decision Tree:
- `max_depth=10` - Profundidad máxima (None = sin límite)
- `min_samples_split=10` - Mínimo muestras para dividir
- `criterion='gini'` - Función de impureza ('gini' o 'entropy')

### Random Forest:
- `n_estimators=100` - Número de árboles (50-500)
- `max_depth=None` - Profundidad (None = sin límite)
- `max_features='sqrt'` - Features por split
- `n_jobs=-1` - Usar todos los cores CPU

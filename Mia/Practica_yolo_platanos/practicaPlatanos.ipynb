```python
# Celda 2: Importar librerías
from ultralytics import YOLO
import os
import matplotlib.pyplot as plt
from PIL import Image
import pandas as pd

print("✅ Librerías importadas correctamente")
```

```python
# Celda 3: Verificar estructura del dataset
RUTA_DATASET = "Banana Ripeness Classification.v2-original-images_modifiedclasses.folder"

# Verificar que existe
if os.path.exists(RUTA_DATASET):
    print(f"✅ Dataset encontrado en: {RUTA_DATASET}")
    
    # Listar contenido
    print(f"\n📁 Contenido:")
    for item in os.listdir(RUTA_DATASET):
        print(f"  - {item}")
else:
    print(f"❌ ERROR: No se encontró el dataset en {RUTA_DATASET}")
    print("Verifica la ruta y que hayas descomprimido el archivo")
```

```python
# Celda 4: Analizar distribución del dataset
def analizar_dataset(ruta_dataset):
    print("=" * 80)
    print("📊 ANÁLISIS DEL DATASET")
    print("=" * 80)
    
    splits = ['train', 'valid', 'test']
    resumen = {}
    
    for split in splits:
        ruta_split = os.path.join(ruta_dataset, split)
        
        if not os.path.exists(ruta_split):
            print(f"⚠️  Carpeta {split} no encontrada")
            continue
        
        print(f"\n📁 {split.upper()}:")
        clases = [d for d in os.listdir(ruta_split) if os.path.isdir(os.path.join(ruta_split, d))]
        
        total_split = 0
        for clase in clases:
            ruta_clase = os.path.join(ruta_split, clase)
            num_imgs = len([f for f in os.listdir(ruta_clase) 
                           if f.lower().endswith(('.jpg', '.jpeg', '.png'))])
            print(f"  - {clase:20}: {num_imgs:4} imágenes")
            total_split += num_imgs
            
            if clase not in resumen:
                resumen[clase] = 0
            resumen[clase] += num_imgs
        
        print(f"  {'TOTAL':20}: {total_split:4} imágenes")
    
    print("\n" + "=" * 80)
    print("📈 RESUMEN TOTAL POR CLASE:")
    print("=" * 80)
    total_general = sum(resumen.values())
    for clase, cantidad in resumen.items():
        porcentaje = (cantidad / total_general * 100) if total_general > 0 else 0
        print(f"  {clase:20}: {cantidad:4} imágenes ({porcentaje:5.1f}%)")
    print(f"  {'TOTAL GENERAL':20}: {total_general:4} imágenes")
    
    return resumen

# Ejecutar análisis
analizar_dataset(RUTA_DATASET)
```

```python
# Celda 5: Entrenar el modelo
print("🏋️  ENTRENAMIENTO DEL MODELO")
print("=" * 80)

# Configuración
MODELO = "yolo11n-seg.pt"  # Puedes cambiar a yolo11s-cls.pt o yolo11m-cls.pt
EPOCHS = 20
IMGSZ = 224
DEVICE = 'cpu'  # Cambiar a 'cuda' si tienes GPU

# Usar archivo data.yaml en lugar de la carpeta
RUTA_DATASET = "data.yaml"

# Cargar modelo
print(f"\n📦 Cargando modelo: {MODELO}")
model = YOLO(MODELO)

# Entrenar
print(f"\n⏱️  Iniciando entrenamiento...")
print(f"  Dataset: {RUTA_DATASET}")
print(f"  Épocas: {EPOCHS}")
print(f"  Device: {DEVICE}")

results = model.train(
    data=RUTA_DATASET,
    epochs=EPOCHS,
    imgsz=IMGSZ,
    device=DEVICE,
    project='banana_training',
    name='ripeness_model',
    plots=True,
    verbose=True
)

print("\n✅ ¡Entrenamiento completado!")
print(f"📁 Resultados en: banana_training/ripeness_model/")
```

```python
# Celda 6: Validar el modelo
print("🔍 VALIDACIÓN DEL MODELO")
print("=" * 80)

# Cargar mejor modelo
best_model = YOLO("banana_training/ripeness_model/weights/best.pt")

# Validar
metrics = best_model.val()

# Mostrar métricas
print(f"\n📊 RESULTADOS:")
print(f"  Top-1 Accuracy: {metrics.top1:.4f} ({metrics.top1*100:.2f}%)")
print(f"  Top-5 Accuracy: {metrics.top5:.4f} ({metrics.top5*100:.2f}%)")
```

```python
# Celda 7: Visualizar matriz de confusión y curvas
print("📊 VISUALIZACIÓN DE RESULTADOS")
print("=" * 80)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Matriz de confusión
confusion_path = "banana_training/ripeness_model/confusion_matrix.png"
if os.path.exists(confusion_path):
    img = Image.open(confusion_path)
    axes[0].imshow(img)
    axes[0].axis('off')
    axes[0].set_title('Matriz de Confusión', fontsize=14, fontweight='bold')
else:
    axes[0].text(0.5, 0.5, 'No encontrada', ha='center', va='center')
    axes[0].axis('off')

# Curvas de entrenamiento
results_path = "banana_training/ripeness_model/results.png"
if os.path.exists(results_path):
    img = Image.open(results_path)
    axes[1].imshow(img)
    axes[1].axis('off')
    axes[1].set_title('Curvas de Entrenamiento', fontsize=14, fontweight='bold')
else:
    axes[1].text(0.5, 0.5, 'No encontrada', ha='center', va='center')
    axes[1].axis('off')

plt.tight_layout()
plt.savefig('resultados/analisis_completo.png', dpi=150, bbox_inches='tight')
plt.show()

print("💾 Gráfica guardada en: resultados/analisis_completo.png")
```

```python
# Celda 8: Realizar predicciones en imágenes nuevas
print("🔮 PREDICCIONES EN IMÁGENES NUEVAS")
print("=" * 80)

def predecir_y_visualizar(ruta_imagen):
    """Predice y muestra resultado"""
    # Cargar modelo
    model = YOLO("banana_training/ripeness_model/weights/best.pt")
    
    # Predecir
    results = model(ruta_imagen)
    probs = results[0].probs
    
    # Obtener clase predicha
    clase_idx = probs.top1
    clase_nombre = model.names[clase_idx]
    confianza = probs.top1conf.item()
    
    # Cargar imagen
    img = Image.open(ruta_imagen)
    
    # Visualizar
    plt.figure(figsize=(10, 6))
    plt.imshow(img)
    plt.axis('off')
    plt.title(f'Predicción: {clase_nombre} ({confianza:.1%})', 
             fontsize=16, fontweight='bold', color='green' if confianza > 0.8 else 'orange')
    plt.tight_layout()
    
    # Guardar
    nombre_archivo = os.path.basename(ruta_imagen).split('.')[0]
    plt.savefig(f'resultados/prediccion_{nombre_archivo}.png', dpi=150, bbox_inches='tight')
    plt.show()
    
    # Mostrar probabilidades
    print(f"\n📸 Imagen: {ruta_imagen}")
    print(f"🏆 Predicción: {clase_nombre}")
    print(f"📊 Confianza: {confianza:.2%}")
    print(f"\n📋 Todas las probabilidades:")
    for idx, prob in enumerate(probs.data):
        print(f"  {model.names[idx]:15}: {prob:.4f} ({prob*100:.1f}%)")
    print("-" * 80)
    
    return clase_nombre, confianza

# Probar con tus imágenes
imagenes_prueba = [
    "imagenes_prueba/banana_verde_1.jpg",
    "imagenes_prueba/banana_madura_1.jpg",
    "imagenes_prueba/banana_pasada_1.jpg",
]

resultados_pruebas = []
for img_path in imagenes_prueba:
    if os.path.exists(img_path):
        clase, conf = predecir_y_visualizar(img_path)
        resultados_pruebas.append({
            'imagen': os.path.basename(img_path),
            'prediccion': clase,
            'confianza': f"{conf:.2%}"
        })
    else:
        print(f"⚠️  No se encontró: {img_path}")

# Resumen
print("\n" + "=" * 80)
print("📋 RESUMEN DE PREDICCIONES")
print("=" * 80)
df_resultados = pd.DataFrame(resultados_pruebas)
print(df_resultados.to_string(index=False))
```

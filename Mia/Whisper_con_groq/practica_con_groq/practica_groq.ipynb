```python
%pip install gtts
from gtts import gTTS

# Texto a convertir
texto = "This is a voice audio for making a test with groq"

# Crear el objeto de audio (en inglés)
tts = gTTS(text=texto, lang='en')

# Guardar el archivo
tts.save("audio_prueba.mp3")
```

```python
import os
from groq import Groq
# Inicializar el cliente de Groq
client = Groq()
# Definir la ruta del archivo de audio
filename = "audio_prueba.mp3"
# Proceso de apertura y transcripción
with open(filename, "rb") as file:
    transcription = client.audio.transcriptions.create(
        file=file,  # Archivo de audio binario
        model="whisper-large-v3-turbo",  # Modelo seleccionado
        prompt="Contexto o palabras clave",  # Opcional para mejorar
        response_format="verbose_json",  # Formato detallado de salida
        language="en",  # Idioma del audio (ISO-639-1)
        temperature=0.0  # Mayor determinismo (0.0)
)
# Resultado final
print(transcription.text)
```

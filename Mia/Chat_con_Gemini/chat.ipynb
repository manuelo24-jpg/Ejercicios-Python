```python
import os 
from dotenv import load_dotenv 
from groq import Groq 
from google import genai 

# Cargar las variables del archivo .env
load_dotenv()

# Inicializar ambos clientes
groq_client = Groq()
gemini_client = genai.Client() # Detecta la clave automáticamente

# Inicializar ambos clientes
groq_client = Groq()
gemini_client = genai.Client() # Detecta la clave automáticamente 
```

```python
# 1. Preguntar al usuario un concepto de programación 
print(" El Comparador Multimodal ")
concepto = "Puedes explicarme los principios de la programacion orientada a objetos?"
```

```python
# 2. Uso de Groq con un rol de sistema sarcástico para una definición rápida 
print("\n[Consultando a Groq]")
respuesta_groq = groq_client.chat.completions.create(
    model="llama-3.3-70b-versatile", # Utilizando el modelo definido en las prácticas anteriores 
    messages=[
        {
            "role": "system", 
            "content": "Eres un programador experto, pero muy sarcástico y gracioso. Define conceptos de programación de forma rápida y con humor." # Definimos la personalidad 
        },
        {
            "role": "user", 
            "content": f"Define rápidamente este concepto: {concepto}"
        }
    ]
)
```

```python
# 3. Uso de Gemini para dar una explicación académica y detallada 
print("[Consultando a Gemini...]\n")
prompt_academico = f"Proporciona una explicación académica, seria y muy detallada sobre el siguiente concepto de programación: '{concepto}'. Incluye ejemplos si es necesario."

try:
    respuesta_gemini = gemini_client.models.generate_content(
        model="gemini-2.5-pro", # El modelo exigido para la tarea final 
        contents=prompt_academico
    )
    # Extraer texto de la respuesta de Gemini si es posible, si no, convertir a str
    if hasattr(respuesta_gemini, "candidates") and respuesta_gemini.candidates:
        respuesta_gemini_text = respuesta_gemini.candidates[0].content
    else:
        respuesta_gemini_text = str(respuesta_gemini)
except Exception as e:
    # Manejo de cuota/errores: usar la respuesta de Groq como alternativa y notificar
    print("Advertencia: fallo al consultar Gemini (posible cuota excedida).", e)
    print("Usando la respuesta de Groq como alternativa.\n")
    respuesta_gemini_text = respuesta_groq.choices[0].message.content

# Mostrar la respuesta final (texto)
print(respuesta_gemini_text)
```

```python
# Mostrar los resultados
print("--- RESPUESTA DE GROQ (SARCÁSTICA Y RÁPIDA) ---")
print(respuesta_groq.choices[0].message.content) 

print("\n--- RESPUESTA DE GEMINI (ACADÉMICA Y DETALLADA) ---")
print(respuesta_gemini_text) 
```

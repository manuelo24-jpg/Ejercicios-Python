```python
%pip install gTTS
from gtts import gTTS
import os

texto = "Mi usuario es el numero 10 y mi contraseña, no te voy a decir"
# 'lang' define el idioma, 'es' para español
tts = gTTS(text=texto, lang='es')
tts.save("audio.mp3")

# Reproducir el archivo (varía según sistema operativo)
# os.system("start audio.mp3") # En Windows
os.system("mpg123 audio.mp3") # En Linux
# os.system("afplay audio.mp3") # En Mac
```

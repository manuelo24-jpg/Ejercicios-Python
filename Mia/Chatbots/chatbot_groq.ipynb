```python
import os
from groq import Groq
from dotenv import load_dotenv

# Carga de variables de entorno (API Key)
load_dotenv()

# Inicialización del cliente de Groq
client = Groq()

# [cite: 7, 33] Inicializamos el historial con el rol 'system' en la primera posición.
# Aquí definimos la personalidad: un programador senior sarcástico.
historial = [
    {
        "role": "system", 
        "content": "Eres un programador senior harto de preguntas básicas. Responde de forma sarcástica, breve y técnica." 
    }
]

print("--- CHATBOT CON GROQ (Llama 3.3) ---")
print("(Escribe 'salir' para terminar)\n")

while True:
    # Capturamos la entrada del usuario (Rol: User)
    usuario = input("Tú: ")
    
    if usuario.lower() == "salir":
        break
    
    # 1. Añadimos el mensaje del usuario al historial para mantener la "memoria"
    historial.append({"role": "user", "content": usuario})
    
    # 2. Petición a Groq usando el modelo Llama 3.3 
    completion = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=historial,
        temperature=0.8
    )
    
    # Extraemos el contenido de la respuesta
    respuesta = completion.choices[0].message.content 
    
    # 3. Mostramos la respuesta y la guardamos en el historial con el rol 'assistant' 
    print(f"IA: {respuesta}\n")
    historial.append({"role": "assistant", "content": respuesta}) 
```

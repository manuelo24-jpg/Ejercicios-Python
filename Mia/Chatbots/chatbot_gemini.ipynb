```python
import os
from google import genai
from dotenv import load_dotenv

# Carga de variables de entorno (API Key)
load_dotenv()

# Inicialización del cliente de Gemini
# El SDK detecta automáticamente la variable GEMINI_API_KEY del archivo .env
client = genai.Client()

# 1. Configuración de la "instrucción del sistema" y parámetros
# En Gemini, la personalidad no es un mensaje en el historial, sino una instrucción base
config_chatbot = {
    "system_instruction": "Eres un programador senior que le encanta ayudar. Responde de forma amable y técnica.",
    "temperature": 1.6  # Un valor equilibrado para un chatbot de propósito general
}

# 2. Creación de la sesión de chat
# Este objeto 'chat' se encargará de gestionar el historial automáticamente
chat = client.chats.create(
    model="gemini-2.5-flash",
    config=config_chatbot
)

print("--- CHATBOT CON GEMINI (2.5 Flash) ---")
print("(Escribe 'salir' para terminar)\n")

while True:
    # Capturamos la entrada del usuario
    usuario = input("Tú: ")
    
    if usuario.lower() == "salir":
        break
    
    # 3. Envío de mensaje a través de la sesión de chat
    # No es necesario gestionar una lista .append(), Gemini lo hace por ti
    response = chat.send_message(usuario)
    
    # Mostramos la respuesta obtenida
    print(f"IA: {response.text}\n")
```
